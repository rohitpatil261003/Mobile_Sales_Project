```python
df['Memory'].fillna(df['Memory'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Storage'].fillna(df['Storage'].mode()[0], inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
df['Memory_GB'] = pd.to_numeric(
    df['Memory'].str.extract(r'(\d+)', expand=False),
    errors='coerce'
)
```

```python
df['Storage_GB'] = pd.to_numeric(
    df['Storage'].str.extract(r'(\d+)', expand=False),
    errors='coerce'
)
```

```python
df.to_csv("Cleaned_Mobile_Data.csv", index=False)
print("Data cleaned successfully! Rows:", len(df))
```

```python
df['Discount'] = df['Original Price'] - df['Selling Price']
```

```python
df['Discount_Percentage'] = (
    df['Discount'] / df['Original Price']
) * 100
```

```python
# Brand count
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    y='Brands',
    order=df['Brands'].value_counts().head(10).index
)

plt.title("Top 10 Mobile Brands")
plt.show()
```

```python
# Price distribution
sns.histplot(df['Selling Price'], kde=True)

plt.title("Mobile Selling Price Distribution")
plt.show()
```

```python
# Rating vs Selling Price
sns.scatterplot(
    data=df,
    x='Rating',
    y='Selling Price'
)

plt.title("Rating vs Selling Price")
plt.show()
```
